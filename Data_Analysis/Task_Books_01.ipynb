### Import Pandas

### Read DataFrame

```python
url = 'https://raw.githubusercontent.com/ozlerhakan/mongodb-json-files/master/datasets/books.json'
books = pd.read_json(url, lines=True)
books
```

### Categorise with page count

```python
books[books['pageCount'] == 0]
```

## Books upto 250 pages

```python
selection_vector = (books['pageCount'] > 0) & (books['pageCount'] < 250)
selection_vector
```

* Books category with pageCount
  * 0 Pages -> "Info Missing"
  * upto 300 -> "Small"
  * 301 to 700 -> "Medium"
  * more than 700 -> "Large

```python
books['bookSize'] = 'Info Missing'  # default

books.loc[(books['pageCount'] > 0) & (books['pageCount'] <= 300), 'bookSize'] = 'Small'
books.loc[(books['pageCount'] > 300) & (books['pageCount'] <= 700), 'bookSize'] = 'Medium'
books.loc[(books['pageCount'] > 700), 'bookSize'] = 'Large'
books['bookSize']
```
