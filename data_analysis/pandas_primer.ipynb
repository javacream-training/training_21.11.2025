### DataFrame als Datencontainer / Collection

* Elementare Berechnungen / Zugriffe

```python
height_series = df['height'] + 1
print(height_series)
```

```python
df['weight']/(df['height']**2) #BMI Berechnung
```

```python
bmi = df['weight']/(df['height']**2)
bmi
```

```python
bmi.info()
```

```python
bmi.describe()
```

* Zuweisung einer Series an einen DataFrame

```python
df['bmi'] = df['weight']/(df['height']**2) #BMI Spalte hinzugefügt
df.info()
```

* Entfernen einer Series aus einem DataFrame

```python
df.drop('GENDER', axis=1)
df.drop(columns=['GENDER'])
```

* Detail: inplace_Operation

```python
df.drop(columns=['GENDER'], inplace=True)
df
```

## Indexzugriff 

```python
df.index[0]
```

```python
slice_df = df.loc[0:2, ['weight', 'height']] #type: DataFrame
slice_df.describe()
```

```python
slice_df = df.iloc[0:2, 3] 
slice_df.describe()
```

### Vergleichsoperatoren

```python
df['height'] > 1.90
```

```python
selection_vector = df['height'] > 1.90
df[selection_vector]
```

```python
tall_people_df = df[df['height'] > 1.90]
tall_people_df
tall_people_df.describe()
```
