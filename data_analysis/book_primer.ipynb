# Aufgabe Books

## Laden des Datensatzes

```python
import pandas as pd
```

```python
books = pd.read_json('https://raw.githubusercontent.com/ozlerhakan/mongodb-json-files/master/datasets/books.json',lines=True)
```

* Was für Spaltennamen haben wir? Welchen Datentyp haben diese?

```python
columns = books.head(0)
print("Spalten: ", list(books.columns))
print("Datentyp der Spalten: ")
for Spalte, Typ in books.dtypes.items():
    print(Spalte, "->", Typ)

```

* Wie viele Datensätze haben wir?

```python
len(books)
```

* Bereinigung unnötiger Spalten

```python
books.drop(columns=['_id', 'publishedDate', 'thumbnailUrl', 'shortDescription', 'longDescription'], inplace=True)
```

```python
slice_page = books['pageCount']
slice_page.describe()
```

* Hinzufügen Kategorie für Seitenanzahl

```python
books['page_category'] = pd.cut(books['pageCount'], bins=[0, 350, 500, float('inf')],labels=['low', 'middle', 'high'])
books.info() 
#Differenz Non-Null Count ergibt sich durch die Bücher mit einer Seitenanzahl von  0 --> Bin beginnt ab 0, ohne diese mit einzuschließen (Alternative wäre -1)
```

* Welche Bücher haben die Kategorie 'Python'?

```python
selection_vector = books['categories'].astype(str).str.contains('[Java]')
selection_vector
```

* Vergleichsoperatoren

```python
number_pages = books['pageCount'] > 400
books[number_pages]
```
